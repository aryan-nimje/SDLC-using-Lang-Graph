```python
from typing import Annotated, TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
```

```python
class State(TypedDict):
    messages : Annotated[list, add_messages]
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    model = "gpt-4.1-mini",
    temperature = 0.5
)
```

```python
from langchain_core.tools import tool
from langgraph.prebuilt import ToolNode
from langgraph.prebuilt import tools_condition
from langgraph.types import Command, interrupt
import datetime
```

```python
from langgraph.checkpoint.memory import MemorySaver

memory = MemorySaver()
config = {"configurable": {"thread_id" : "1"}}
```

```python
import sys
from pathlib import Path

# Go up one directory from "testing grounds" to the project root
project_root = Path.cwd().parent

sys.path.append(str(project_root))
```

```python
from tools.tester import critique_file, read_workspace
```

```python
tools = [critique_file, read_workspace]
```

```python
tester = llm.bind_tools(tools = tools)

def testing_agent(state:State):
    response = tester.invoke(state["messages"])
    return {"messages": [response]}
```

```python
graph_builder = StateGraph(State)

graph_builder.add_node("Tester", testing_agent)
graph_builder.add_node("tools", ToolNode(tools=tools))

graph_builder.add_edge(START, "Tester")
graph_builder.add_conditional_edges(
    "Tester",
    tools_condition
)
graph_builder.add_edge("tools", "Tester")

graph = graph_builder.compile(checkpointer=memory)
```
